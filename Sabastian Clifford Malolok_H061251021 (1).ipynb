# PROMPT 3 — PEMPROSESAN DATA SEISMIK REFLEKSI (PENGURANGAN NOISE GROUND ROLL)

## 1. Narasi Permasalahan

Pemprosesan data seismik memerlukan alir kerja (workflow) penyaringan yang teliti supaya *ground roll* dapat dihilangkan tanpa menjejaskan isyarat pantulan utama (primary reflections). AI dapat membantu mereka bentuk strategi tapis (filtering), menentukan parameter tapis yang optimum, dan mengevaluasi spektrum frekuensi sebelum dan selepas pemprosesan.

## 2. Struktur Good Prompt

### T — Task

### C — Context
Data seismik pantulan darat dirakam menggunakan geofon julat frekuensi luas. Rekod shot menunjukkan gangguan ground roll yang dominan pada frekuensi rendah (1–15 Hz) dan halaju rendah, menutupi horizon pantulan pada masa 0.2 hingga 1.0 saat.

### I — Input
Data yang diberikan berupa:
1. Fail rekod seismik (shot gather) dalam format SEG-Y atau array data masa-jarak.

3. Maklumat jarak offset geofon dari sumber (shotpoint).

### K — Constraints
1. Jangan gunakan tapis yang terlalu agresif sehingga memotong frekuensi isyarat pantulan cetek.
2. Jelaskan perbezaan antara tapis F-K (Frequency-Wavenumber) dan tapis Bandpass yang dicadangkan.
3. Jangan mengubah fasa asal isyarat tanpa sebab yang munasabah.
4. Tunjukkan analisis spektrum sebelum dan selepas penapisan.

### O — Output
Hasil harus mencakup:
1. Carta alir (workflow) pemprosesan penapisan ground roll.
2. Cadangan parameter julat frekuensi (cut-off frequency) dan parameter tapis F-K.
3. Perbandingan paparan shot gather sebelum dan selepas penyaringan.
4. Analisis spektrum frekuensi (Amplitude-Frequency Spectrum).

### Q — Quality Criteria

Dalam sistem geoterma, data MT digunakan untuk memetakan lapisan penudung konduktif (*cap rock* / zon tanah liat smektit) yang mempunyai kerintangan rendah, serta zon takungan geoterma di bawahnya. AI boleh membantu memproses keluk kerintangan semak dan fase, menilai Kedalaman Kulit (*Skin Depth*), dan membina model konseptual geoterma berdasarkan taburan kerintangan.

Survei MT dilakukan di kawasan prospek geoterma gunung api. Data yang diproses menghasilkan keluk kerintangan semak dan fase terhadap frekuensi. Frekuensi tinggi mencerminkan kedalaman cetek, manakala frekuensi rendah mencerminkan struktur dalam.

### V — Verification
Semak semula konsistensi antara keluk kerintangan semak dan keluk fase (mengikut hubungan Kramers-Kronig), dan pastikan anggaran kedalaman cap rock selari dengan data manifestasi permukaan yang wujud.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Menyiapkan Data Sintetis (X, Z, Apparent Resistivity)
np.random.seed(42)
x = np.linspace(0, 100, 20)
z = np.linspace(0, 30, 10)
X, Z = np.meshgrid(x, z)

# Simulasi distribusi kerintangan semak (Ohm-m)
# Nilai rendah (<15): Akuifer/Batuan berair, Sedang (15-100): Batuan lapuk, Tinggi (>100): Bedrock/Batuan Kering
res_true = 100 - 80 * np.exp(-((X - 50)**2 + (Z - 15)**2) / 200) + np.random.normal(0, 2, X.shape)

df_res = pd.DataFrame({
    'X_m': X.flatten(),
    'Depth_m': Z.flatten(),
    'Apparent_Resistivity_Ohm_m': res_true.flatten()
})

# 2. Pengelompokan Litologi & Kriteria Julat Kerintangan
def classify_litology(rho):
    if rho < 20:
        return 'Zon Berair / Akuifer'
    elif 20 <= rho <= 80:
        return 'Batuan Lapuk / Tanah Lemak'
    else:
        return 'Batuan Dasar (Bedrock) / Batuan Kering'

df_res['Litologi_Teragak'] = df_res['Apparent_Resistivity_Ohm_m'].apply(classify_litology)

# 3. Visualisasi Profil Kerintangan 2D
plt.figure(figsize=(10, 5))
contour = plt.contourf(X, Z, res_true, levels=15, cmap='jet_r')
plt.colorbar(contour, label='Kerintangan Semak (Ohm-m)')
plt.gca().invert_yaxis()
plt.title('Profil Kerintangan Elektrik 2D (Konfigurasi Wenner-Schlumberger)')
plt.xlabel('Jarak Lintasan X (m)')
plt.ylabel('Kedalaman Z (m)')

# Penandaan Potensi Akuifer & Titik Gerudi
plt.plot(50, 15, 'r*', markersize=15, label='Cadangan Titik Gerudi (X=50m, Z=15m)')
plt.legend(loc='lower right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# Ringkasan Output
print("--- TABEL RINGKASAN JULAT KERINTANGAN ---")
print(df_res.groupby('Litologi_Teragak')['Apparent_Resistivity_Ohm_m'].agg(['min', 'max', 'mean']))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# 1. Menyiapkan Data Observasi Sintetis (Sesar)
x_obs = np.linspace(-1000, 1000, 40)
# Respon analitis teoritis untuk struktur sesar/step
g_obs_true = 10 + 15 * (0.5 + (1/np.pi) * np.arctan(x_obs / 200)) 
noise = np.random.normal(0, 0.5, size=x_obs.shape)
g_obs = g_obs_true + noise  # Data observasi dengan noise

# 2. Fungsi Pemodelan Kedepan (Forward Modeling Sesar 2D)
def forward_fault_model(x, params):
    # params: [g_background, delta_g, depth, x_center]
    g_bg, delta_g, z0, x0 = params
    return g_bg + delta_g * (0.5 + (1/np.pi) * np.arctan((x - x0) / z0))

# 3. Fungsi Objective (Fungsi Misfit / Error)
def objective_function(params):
    g_calc = forward_fault_model(x_obs, params)
    return np.sqrt(np.mean((g_obs - g_calc)**2))  # RMS Error

# 4. Inversi Non-Linear
initial_guess = [5, 10, 100, 0]
bounds = [(0, 20), (1, 50), (10, 500), (-500, 500)]

result = minimize(objective_function, initial_guess, method='L-BFGS-B', bounds=bounds)
g_calc_opt = forward_fault_model(x_obs, result.x)

# Penghitungan RMS Error (%)
rms_error = (result.fun / np.std(g_obs)) * 100

# 5. Visualisasi Hasil Inversi
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)

# Graf Profil Anomali Graviti
ax1.plot(x_obs, g_obs, 'ko', label='Anomali Observasi')
ax1.plot(x_obs, g_calc_opt, 'r-', linewidth=2, label=f'Anomali Kalkulasi (RMS: {rms_error:.2f}%)')
ax1.set_ylabel('Anomali Bouguer (mGal)')
ax1.set_title('Pemodelan Inversi Anomali Graviti 2D Structure Sesar')
ax1.legend()
ax1.grid(True)

# Visualisasi Model Sesar 2D
X_grid, Z_grid = np.meshgrid(x_obs, np.linspace(0, 400, 20))
density_model = np.zeros_like(X_grid)
density_model[X_grid > result.x[3]] = 0.3  # Kontras ketumpatan (g/cc)

im = ax2.contourf(X_grid, Z_grid, density_model, cmap='viridis')
ax2.axvline(x=result.x[3], color='red', linestyle='--', label=f'Perkiraan Sesar (X={result.x[3]:.1f}m)')
ax2.set_ylim(400, 0)
ax2.set_xlabel('Jarak Lintasan (m)')
ax2.set_ylabel('Kedalaman (m)')
ax2.legend()
plt.colorbar(im, ax=ax2, label='Kontras Ketumpatan (g/cm³)')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, filtfilt

# 1. Menyiapkan Data Sintetis Shot Gather
dt = 0.002  # Sampling rate 2 ms
t = np.arange(0, 1.0, dt)
offsets = np.linspace(10, 200, 40)
T, X = np.meshgrid(t, offsets)

# Sinyal Pantulan Utuh (Primary Reflection - High Frequency)
ref1 = np.exp(-((T - (0.3 + 0.0005*X)) / 0.02)**2) * np.sin(2 * np.pi * 35 * T)
ref2 = np.exp(-((T - (0.6 + 0.0008*X)) / 0.02)**2) * np.sin(2 * np.pi * 30 * T)
reflection_signal = ref1 + ref2

# Noise Ground Roll (Low Frequency 5-10Hz, Low Velocity)
ground_roll = 3.0 * np.exp(-((T - X/150) / 0.08)**2) * np.sin(2 * np.pi * 8 * T)

# Combined Shot Gather Data
raw_data = reflection_signal + ground_roll

# 2. Perancangan Penapis Bandpass (Butterworth Filter)
def bandpass_filter(data, lowcut=18.0, highcut=80.0, fs=500.0, order=4):
    nyq = 0.5 * fs
    low = lowcut / nyq
    high = highcut / nyq
    b, a = butter(order, [low, high], btype='band')
    return filtfilt(b, a, data, axis=1)

filtered_data = bandpass_filter(raw_data, fs=1/dt)

# 3. Analisis Spektrum Amplitudo-Frekuensi
fft_raw = np.abs(np.fft.rfft(raw_data[0]))
fft_filt = np.abs(np.fft.rfft(filtered_data[0]))
freqs = np.fft.rfftfreq(len(t), dt)

# 4. Visualisasi Hasil Pemprosesan
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

# Raw Shot Gather
axes[0].imshow(raw_data.T, aspect='auto', cmap='seismic', extent=[10, 200, 1.0, 0])
axes[0].set_title('Raw Shot Gather (Ada Ground Roll)')
axes[0].set_ylabel('Masa (s)')
axes[0].set_xlabel('Offset (m)')

# Filtered Shot Gather
axes[1].imshow(filtered_data.T, aspect='auto', cmap='seismic', extent=[10, 200, 1.0, 0])
axes[1].set_title('Filtered Shot Gather (Bandpass 18-80Hz)')
axes[1].set_xlabel('Offset (m)')

# Spektrum Frekuensi
axes[2].plot(freqs, fft_raw, 'r', label='Sebelum Filter')
axes[2].plot(freqs, fft_filt, 'b', label='Selepas Filter')
axes[2].set_xlim(0, 100)
axes[2].set_title('Analisis Spektrum Frekuensi')
axes[2].set_xlabel('Frekuensi (Hz)')
axes[2].set_ylabel('Amplitudo')
axes[2].legend()
axes[2].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Menyiapkan Data Well Log Sintetis (1200m - 1500m)
depth = np.linspace(1200, 1500, 300)

# Simulasi Bacaan Log
gr = 60 + 30 * np.sin(depth / 10) + np.random.normal(0, 5, len(depth)) # API
rd = 10 ** (0.5 + 1.2 * np.cos(depth / 15) + np.random.normal(0, 0.1, len(depth))) # Ohm-m
rhob = 2.4 - 0.2 * np.cos(depth / 15) + np.random.normal(0, 0.02, len(depth)) # g/cc
nphi = 0.2 + 0.15 * np.cos(depth / 15) + np.random.normal(0, 0.02, len(depth)) # v/v

df_well = pd.DataFrame({'Depth': depth, 'GR': gr, 'RD': rd, 'RHOB': rhob, 'NPHI': nphi})

# 2. Penghitungan Petrofisika (Persamaan Archie Standard)
# Parameters: a=1, m=2, n=2, Rw=0.05 Ohm-m, Matrix Density = 2.65 g/cc (Sandstone)
rho_matrix = 2.65
rho_fluid = 1.0

# Keterliangan Total / Porositas (\phi)
df_well['Porosity'] = (rho_matrix - df_well['RHOB']) / (rho_matrix - rho_fluid)
df_well['Porosity'] = df_well['Porosity'].clip(lower=0.001, upper=0.4)

# Ketepuan Air (Sw) Archie: Sw = sqrt((a * Rw) / (phi^m * Rt))
a, m, n, Rw = 1.0, 2.0, 2.0, 0.05
df_well['Sw'] = np.sqrt((a * Rw) / ((df_well['Porosity']**m) * df_well['RD']))
df_well['Sw'] = df_well['Sw'].clip(lower=0.0, upper=1.0)

# 3. Kriteria Net Pay (GR < 60, Sw < 0.5, Porosity > 0.12)
df_well['Is_Pay'] = (df_well['GR'] < 60) & (df_well['Sw'] < 0.5) & (df_well['Porosity'] > 0.12)

# 4. Visualisasi Log
fig, axes = plt.subplots(1, 4, figsize=(12, 8), sharey=True)

# Track 1: Gamma Ray
axes[0].plot(df_well['GR'], df_well['Depth'], 'g-')
axes[0].axvline(60, color='gray', linestyle='--')
axes[0].set_xlabel('GR (API)')
axes[0].set_title('Track 1: GR')

# Track 2: Resistivity
axes[1].semilogx(df_well['RD'], df_well['Depth'], 'r-')
axes[1].set_xlabel('RD (Ohm-m)')
axes[1].set_title('Track 2: Deep Res')

# Track 3: NPHI-RHOB Crossover
axes[2].plot(df_well['NPHI'], df_well['Depth'], 'b-', label='NPHI')
axes2_twin = axes[2].twiny()
axes2_twin.plot(df_well['RHOB'], df_well['Depth'], 'r-', label='RHOB')
axes[2].set_xlabel('NPHI (v/v)', color='b')
axes2_twin.set_xlabel('RHOB (g/cc)', color='r')
axes[2].set_title('Track 3: Porosity Logs')

# Track 4: Net Pay Zone
axes[3].fill_betweenx(df_well['Depth'], 0, df_well['Is_Pay'].astype(int), color='gold', alpha=0.7)
axes[3].set_xlabel('Pay Zone (1=Yes)')
axes[3].set_title('Track 4: Net Pay')

axes[0].set_gca().invert_yaxis()
axes[0].set_ylabel('Kedalaman (m)')
plt.tight_layout()
plt.show()

print(f"Total Net Pay Thickness: {df_well['Is_Pay'].sum() * (depth[1]-depth[0]):.2f} m")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Data Sintetis Pengukuran MT
freq = np.logspace(3, -2, 30) # Frekuensi dari 1000 Hz ke 0.01 Hz
rho_a = np.zeros_like(freq)
phase = np.zeros_like(freq)

# Model Konseptual Geoterma 1D:
# High Freq (Cetek) -> Penutup (Cap Rock/Smektit): Rho Rendah (<10 Ohm-m)
# Med Freq (Sederhana) -> Reservoir: Rho Sedang (10-60 Ohm-m)
# Low Freq (Dalam) -> Deep Conductor / Sumber Haba: Rho Rendah
for i, f in enumerate(freq):
    if f > 50:
        rho_a[i] = 40 + np.random.normal(0, 2)
        phase[i] = 45
    elif 1 <= f <= 50: # Cap Rock Zone
        rho_a[i] = 5 + np.random.normal(0, 0.5)
        phase[i] = 65
    else: # Geothermal Reservoir & Deep Source
        rho_a[i] = 35 + np.random.normal(0, 2)
        phase[i] = 30

# 2. Penghitungan Kedalaman Penetrasi (Skin Depth: delta = 503 * sqrt(rho / f))
skin_depth_m = 503 * np.sqrt(rho_a / freq)

df_mt = pd.DataFrame({
    'Frequency_Hz': freq,
    'Apparent_Resistivity_Ohm_m': rho_a,
    'Phase_deg': phase,
    'Skin_Depth_m': skin_depth_m
})

# 3. Plot Kurva MT (Resistivity & Phase vs Frequency)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 7), sharex=True)

# Graf Kerintangan Semak
ax1.loglog(df_mt['Frequency_Hz'], df_mt['Apparent_Resistivity_Ohm_m'], 'bo-', label='Apparent Rho')
ax1.set_ylabel('Apparent Resistivity (Ohm-m)')
ax1.set_title('Kurva Respon Magnetotellurik (MT) - Sistem Geoterma')
ax1.grid(True, which="both", ls="--")
ax1.axhspan(0, 10, color='red', alpha=0.2, label='Lapisan Cap Rock (<10 Ohm-m)')
ax1.legend()

# Graf Fase
ax2.semilogx(df_mt['Frequency_Hz'], df_mt['Phase_deg'], 'ro-', label='Phase')
ax2.set_xlabel('Frekuensi (Hz)')
ax2.set_ylabel('Fase (derajat)')
ax2.grid(True, which="both", ls="--")
ax2.legend()

plt.gca().invert_xaxis() # Frekuensi tinggi di kiri (cetek), frekuensi rendah di kanan (dalam)
plt.tight_layout()
plt.show()

# Display Ringkasan Zonasi Kedalaman
print("--- ESTIMASI ZONASI KEDALAMAN (SKIN DEPTH) ---")
cap_rock_depth = df_mt[df_mt['Apparent_Resistivity_Ohm_m'] < 10]['Skin_Depth_m']
print(f"Perkiraan Kedalaman Cap Rock: {cap_rock_depth.min():.1f} m hingga {cap_rock_depth.max():.1f} m")
```
